```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from src.data.validation import validate_dataset_directory
from src.data.loader import create_datasets
from utils.file_utils import load_yaml_config
```

```python
cfg = load_yaml_config("configs/config.yaml")
report = validate_dataset_directory("data", "reports/dataset_report.json", "reports/dataset_summary.csv")
print("Dataset Health Status:", report.get("is_healthy", True))
print("Total Valid Images:", report.get("total_images", 0))
print("Class Counts:", report.get("class_counts", {}))
```

```python
if "class_counts" in report and report["class_counts"]:
    df_counts = pd.DataFrame(list(report["class_counts"].items()), columns=["Class", "Count"])
    plt.figure(figsize=(10, 4), facecolor="#FFFFFF")
    sns.barplot(data=df_counts, x="Class", y="Count", palette="Blues_d")
    plt.title("Class Distribution across Manufacturing Defects", fontsize=13, fontweight="bold")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()
```
